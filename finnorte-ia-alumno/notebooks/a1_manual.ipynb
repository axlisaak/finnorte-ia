```python
import numpy as np

# Dimensiones: [base_datos, latencia, seguridad, despliegue]
# F1 y F2 comparten el tema de caída de Base de Datos. F3 es un tema de Despliegue/Seguridad.

F1 = np.array([0.9, 0.2, 0.0, 0.0])  # "La base de datos rechazó todas las conexiones nuevas"
F2 = np.array([0.85, 0.3, 0.05, 0.0]) # "El pod de PostgreSQL se reinició por falta de memoria"
F3 = np.array([0.0, 0.1, 0.8, 0.9])  # "El despliegue falló porque el certificado SSL expiró"

print("--- 1. Vectores Originales ---")
print(f"F1 (Caída BD): {F1}")
print(f"F2 (Reinicio BD): {F2}")
print(f"F3 (Fallo SSL): {F3}\n")

# --- 2. Normalización L2 ---
def normalizar(v):
    norma = np.linalg.norm(v)
    return v / norma if norma != 0 else v

F1_norm = normalizar(F1)
F2_norm = normalizar(F2)
F3_norm = normalizar(F3)

print("--- 2. Vectores Normalizados ---")
print(f"F1 normalizado: {F1_norm.round(4)}")
print(f"F2 normalizado: {F2_norm.round(4)}")
print(f"F3 normalizado: {F3_norm.round(4)}\n")

# --- 3. Similitud Coseno ---
def similitud_coseno(v1, v2):
    # Como ya están normalizados, el coseno es solo el producto punto
    return np.dot(v1, v2)

cos_F1_F2 = similitud_coseno(F1_norm, F2_norm)
cos_F1_F3 = similitud_coseno(F1_norm, F3_norm)

print("--- 3. Similitud Coseno ---")
print(f"Similitud F1-F2 (Mismo tema - BD): {cos_F1_F2:.4f}")
print(f"Similitud F1-F3 (Temas distintos): {cos_F1_F3:.4f}")
```
